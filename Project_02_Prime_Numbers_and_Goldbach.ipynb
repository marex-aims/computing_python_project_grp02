## The problem

A **prime number** has only two divisors: 1 and itself. Examples: 2, 3, 5, 7, 11, 13. The number 1 is **not** prime.

**Goldbach's idea:** every even number bigger than 2 is the sum of two primes.

**Example:** 28 = 5 + 23 and 28 = 11 + 17. Nobody has found an even number where this is false.

**How to test if `n` is prime:** `n` is prime if `n > 1` and no number `d` from 2 to `n - 1` divides `n`. In Python, `n % d == 0` means that `d` divides `n`.

## Tasks

**Task 1.** Write `is_prime(n)`. It returns `True` or `False`.

Example: `is_prime(97)` gives `True`, `is_prime(91)` gives `False` (because 91 = 7 x 13), `is_prime(1)` gives `False`.

**Task 2.** Write `primes_up_to(n)`. It returns the list of all primes from 2 to `n`.

Example: `primes_up_to(20)` gives `[2, 3, 5, 7, 11, 13, 17, 19]`.

How many primes are there up to 100?

**Task 3.** Write `goldbach_pairs(n)`. It returns the list of pairs `(p, q)` where `p` and `q` are prime, `p <= q`, and `p + q = n`.

Example: `goldbach_pairs(10)` gives `[(3, 7), (5, 5)]`.

**Task 4.** Look at the even numbers 4, 6, 8, ..., 500. Find the number that has the **most** pairs. Print this number and its number of pairs. (If two numbers have the same count, keep the smaller one.)

Example: for limit 100, the answer is 90 with 9 pairs.

## Your code

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:            
            return False
    return True

print('25:', is_prime(25))  
print('91:', is_prime(91))  
print('1:', is_prime(1))    
print('2:', is_prime(2))    
```

```python
def primes_up_to(n):
    return [i for i in range(2, n + 1) if is_prime(i)]

print('Primes up to 25:', primes_up_to(25))
print('How many primes up to 100?', len(primes_up_to(100)))
```

```python
def goldbach_pairs(n):
    if n < 4 or n % 2 != 0:
        return []
    primes = primes_up_to(n)
    pairs = []
    for p in primes:
        if (n - p) in primes and p <= n - p:
            pairs.append((p, n - p))
    return pairs

print('Goldbach(10):', goldbach_pairs(10))
print('Goldbach(28):', goldbach_pairs(28))
```

```python
def find_best(limit):
    best_number = None
    best_count = -1
    for n in range(4, limit + 1, 2):
        count = len(goldbach_pairs(n))
        if count > best_count:
            best_number = n
            best_count = count
    return (best_number, best_count) if best_number is not None else (None, 0)

for limit in (100, 200, 500):
    print(f'Limit {limit}: {find_best(limit)}')
```

## Extra : (our solution )

A twin prime pair is `(p, p+2)` such that **both** numbers are prime and **both** are strictly below 100. For example, `(11,13)` qualifies, but `(11,17)` does not (difference 6).

**Step-by-step:** (1) get primes below 100, (2) consider each `p`, (3) check if `p+2` belongs to the prime collection, (4) append exactly one tuple. Since the smaller prime `p` uniquely identifies the pair, there are no reversed duplicates.

Expected eight pairs: `(3,5), (5,7), (11,13), (17,19), (29,31), (41,43), (59,61), (71,73)`.

```python
def twin_primes_below(limit):
    primes = primes_up_to(limit - 1)
    prime_set = set(primes)
    return [(p, p + 2) for p in primes if p + 2 in prime_set]

twins = twin_primes_below(100)
print('Twin primes below 100:', twins)
print('Number of twin prime pairs:', len(twins))
```
